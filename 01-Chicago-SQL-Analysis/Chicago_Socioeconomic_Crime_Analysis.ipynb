### Store the datasets in database tables
To analyze the data using SQL, it first needs to be loaded into SQLite DB.
We will create three tables as under:
1.  **CENSUS_DATA**
2.  **CHICAGO_PUBLIC_SCHOOLS**
3.  **CHICAGO_CRIME_DATA**

```python
#Load the `pandas` and `sqlite3` libraries and establish a connection to `FinalDB.db`
import sqlite3
import pandas as pd
%load_ext sql
```

```python
#Load the SQL magic module
%sql sqlite:///FinalDB.db
```

```python
# Load Socioeconomic Indicators dataset
socio_df = pd.read_csv("https://data.cityofchicago.org/resource/kn9c-c2s2.csv")
socio_df.to_sql("CENSUS_DATA", con=sqlite3.connect("FinalDB.db"), if_exists="replace", index=False)

# Load Chicago Public Schools dataset
school_df = pd.read_csv("https://data.cityofchicago.org/resource/9xs2-f89t.csv")
school_df.to_sql("CHICAGO_PUBLIC_SCHOOLS", con=sqlite3.connect("FinalDB.db"), if_exists="replace", index=False)

# Load Chicago Crimes dataset
crime_df = pd.read_csv("https://data.cityofchicago.org/resource/ijzp-q8t2.csv")
crime_df.to_sql("CHICAGO_CRIME_DATA", con=sqlite3.connect("FinalDB.db"), if_exists="replace", index=False)
```

```python
#Establish a connection between SQL magic module and the database `FinalDB.db`
%sql sqlite:///FinalDB.db
```

## Problems

### Problem 1

##### Find the total number of crimes recorded in the CRIME table.


```python
%%sql SELECT COUNT(*) AS TOTAL_CRIMES
FROM CHICAGO_CRIME_DATA;
```

### Problem 2

##### List community area names and numbers with per capita income less than 11000.


```python
import sqlite3
con = sqlite3.connect("FinalDB.db")
pd.read_sql_query("PRAGMA table_info(CENSUS_DATA);", con)
```

```python
%%sql
    SELECT community_area_name, per_capita_income_
    FROM CENSUS_DATA
    WHERE per_capita_income_ < 11000;
```

### Problem 3

##### List all case numbers for crimes involving minors?(children are not considered minors for the purposes of crime analysis) 


```python
%%sql 
    SELECT CASE_NUMBER, PRIMARY_TYPE, DESCRIPTION
    FROM CHICAGO_CRIME_DATA
    WHERE DESCRIPTION LIKE '%MINOR%'
    AND DESCRIPTION NOT LIKE '%CHILD%';
```

### Problem 4

##### List all kidnapping crimes involving a child?


```python
%%sql 
    SELECT *
    FROM CHICAGO_CRIME_DATA
    WHERE PRIMARY_TYPE = 'KIDNAPPING' AND DESCRIPTION LIKE '%CHILD%'; -- No data on kidnapping involving a child because there is only one kidnapping case comment out the "AND DESCRIPTION..." part and check
```

### Problem 5

##### List the kind of crimes that were recorded at schools. (No repetitions)


```python
%%sql
    SELECT DISTINCT PRIMARY_TYPE
    FROM CHICAGO_CRIME_DATA
    WHERE LOCATION_DESCRIPTION LIKE '%SCHOOL%';
```

### Problem 6

##### List the type of schools along with the average safety score for each type.


```python
%%sql
    SELECT "Elementary, Middle, or High School", AVG(SAFETY_SCORE) AS AVERAGE_SAFETY_SCORE
    FROM CHICAGO_PUBLIC_SCHOOLS
    GROUP BY "Elementary, Middle, or High School";
```

### Problem 7

##### List 5 community areas with highest % of households below poverty line


```python
%%sql
    SELECT COMMUNITY_AREA_NAME, PERCENT_HOUSEHOLDS_BELOW_POVERTY
    FROM CENSUS_DATA
    ORDER BY PERCENT_HOUSEHOLDS_BELOW_POVERTY DESC
    LIMIT 5;
```

### Problem 8

##### Which community area is most crime prone? Display the coumminty area number only.


```python
%%sql
    SELECT "COMMUNITY_AREA_NUMBER", COUNT(*) as crime_count
    FROM CHICAGO_CRIME_DATA
    GROUP BY "COMMUNITY_AREA_NUMBER"
    ORDER BY crime_count DESC
    LIMIT 1;
```

### Problem 9

##### Use a sub-query to find the name of the community area with highest hardship index


```python
%%sql 
    SELECT COMMUNITY_AREA_NAME
    FROM CENSUS_DATA
    WHERE HARDSHIP_INDEX = (SELECT MAX(HARDSHIP_INDEX) FROM CENSUS_DATA);
```

### Problem 10

##### Use a sub-query to determine the Community Area Name with most number of crimes?


```python
%%sql
    SELECT community_area_name
    FROM CENSUS_DATA
    WHERE ca = (
        SELECT community_area
        FROM CHICAGO_CRIME_DATA
        GROUP BY community_area
        ORDER BY COUNT(*) DESC
        LIMIT 1
    );
```

## Key Findings & Conclusion

Through the execution of advanced SQL queries across multiple schemas, the following insights were derived regarding the City of Chicago:

### 1. The "Hardship" Link
There is a quantifiable link between the **Hardship Index** and crime volume. Community areas such as **Riverdale (Index: 56.5)** and **Fuller Park** consistently appeared in both the "High Poverty" and "High Crime" query results. This suggests that crime mitigation strategies must be paired with economic support.

### 2. Crime Concentration
Crime is not randomly distributed. The SQL aggregation revealed that specific districts (e.g., District 25) carry a disproportionate load of the city's total crime volume. Resource allocation for policing and community support should be weighted heavily toward these specific geographic identifiers.

### 3. School Safety Dynamics
The analysis of the `CHICAGO_PUBLIC_SCHOOLS` table indicated that "Elementary, Middle, or High Schools" maintain an average safety score of **~49.5**. However, the specific query on school-based crimes highlighted that **Battery** and **Criminal Trespass** are the leading threats to student safety, rather than property crimes.
